```python
# Student Performance Analysis Using Python

print("STUDENT PERFORMANCE ANALYSIS")
print("-----------------------------")
print("This project analyzes student performance.")
print("We will study marks, attendance, and study hours.")
```

```python
import pandas as pd

data = {
    "Student": ["Asha", "Rahul", "Priya", "Kiran", "Sneha",
                "Arjun", "Divya", "Rohan", "Meena", "Vikram"],
    "Study_Hours": [2, 5, 6, 3, 8, 4, 7, 2, 9, 5],
    "Attendance": [65, 85, 90, 70, 95, 75, 92, 60, 98, 82],
    "Assignment_Average": [60, 80, 88, 65, 95, 72, 90, 55, 98, 78],
    "Final_Marks": [55, 78, 85, 60, 95, 70, 89, 50, 98, 76]
}

df = pd.DataFrame(data)

print("Student Dataset Created Successfully!")
display(df)
```

```python
# Step 4: Data Cleaning

print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates().copy()

print("\nDataset after removing duplicates:")
display(df)

print("Data cleaning completed!")
```

```python

# Step 5: Statistical Analysis

print("STUDENT PERFORMANCE SUMMARY")
print("----------------------------")

print("Average Final Marks:", df["Final_Marks"].mean())
print("Highest Final Marks:", df["Final_Marks"].max())
print("Lowest Final Marks:", df["Final_Marks"].min())

print("Average Study Hours:", df["Study_Hours"].mean())
print("Average Attendance:", df["Attendance"].mean())

print("\nComplete Statistical Summary:")
display(df.describe())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Marks",
    s=100
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours per Day")
plt.ylabel("Final Marks")
plt.grid(True)

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Final_Marks",
    s=100
)

plt.title("Attendance vs Final Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")
plt.grid(True)

plt.show()
```

```python

# Correlation between numerical variables

correlation = df[
    ["Study_Hours", "Attendance",
     "Assignment_Average", "Final_Marks"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance Correlation Heatmap")
plt.show()
```

```python

# Step 9: Correlation Analysis

correlation_with_marks = df[
    ["Study_Hours", "Attendance",
     "Assignment_Average", "Final_Marks"]
].corr()["Final_Marks"].drop("Final_Marks")

print("Correlation with Final Marks:")
print(correlation_with_marks.sort_values(ascending=False))

strongest_factor = correlation_with_marks.abs().idxmax()

print("\nFactor most strongly related to Final Marks:")
print(strongest_factor)
```
